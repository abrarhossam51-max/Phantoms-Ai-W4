```python
import pandas as pd

from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

df = pd.read_csv("train.csv")

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])


df = df.drop(columns=["Cabin", "Name", "Ticket"])


df = pd.get_dummies(
    df,
    columns=["Sex", "Embarked"],
    drop_first=True
)


X = df.drop(columns=["Survived"])
y = df["Survived"]


cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


logistic_model = LogisticRegression(max_iter=1000)
tree_model = DecisionTreeClassifier(random_state=42)


logistic_scores = cross_val_score(
    logistic_model,
    X,
    y,
    cv=cv,
    scoring="accuracy"
)


tree_scores = cross_val_score(
    tree_model,
    X,
    y,
    cv=cv,
    scoring="accuracy"
)


logistic_mean = logistic_scores.mean()
logistic_std = logistic_scores.std()

tree_mean = tree_scores.mean()
tree_std = tree_scores.std()


print("Logistic Regression")
print("Scores:", logistic_scores)
print("Mean Accuracy:", logistic_mean)
print("Standard Deviation:", logistic_std)

print("\nDecision Tree")
print("Scores:", tree_scores)
print("Mean Accuracy:", tree_mean)
print("Standard Deviation:", tree_std)
```
